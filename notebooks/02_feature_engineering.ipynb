## Objectives

The main objectives of this notebook are:

- Load the cleaned HR dataset.
- Remove features that do not contribute to prediction.
- Encode the target variable (Attrition).
- Convert categorical variables into numerical values.
- Save the processed dataset for model training.

```python
# ==========================================================
# FEATURE ENGINEERING PIPELINE
# ==========================================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder


# ----------------------------------------------------------
# 1. Load cleaned dataset
# ----------------------------------------------------------

df = pd.read_csv("../data/processed/employee_attrition_clean.csv")

print("Original shape:", df.shape)


# ----------------------------------------------------------
# 2. Check missing values
# ----------------------------------------------------------

print("\nMissing values:")
print(df.isnull().sum().sum())


# ----------------------------------------------------------
# 3. Remove irrelevant features
# ----------------------------------------------------------

columns_to_drop = [
    "EmployeeCount",
    "EmployeeNumber",
    "Over18",
    "StandardHours"
]

df = df.drop(columns=columns_to_drop)

print("\nAfter removing useless columns:")
print(df.shape)


# ----------------------------------------------------------
# 4. Encode target variable
# ----------------------------------------------------------

df["Attrition"] = df["Attrition"].map({
    "No": 0,
    "Yes": 1
})


# ----------------------------------------------------------
# 5. Identify categorical variables
# ----------------------------------------------------------

categorical_columns = df.select_dtypes(
    include="object"
).columns

print("\nCategorical features:")
print(list(categorical_columns))


# ----------------------------------------------------------
# 6. Encode categorical variables
# ----------------------------------------------------------

encoder = LabelEncoder()

for column in categorical_columns:
    df[column] = encoder.fit_transform(df[column])


# ----------------------------------------------------------
# 7. Check final dataset
# ----------------------------------------------------------

print("\nFinal dataset information:")
df.info()


# ----------------------------------------------------------
# 8. Save feature engineered dataset
# ----------------------------------------------------------

df.to_csv(
    "../data/processed/hr_feature_engineered.csv",
    index=False
)

print("\nFeature engineering completed successfully!")
print("Saved: hr_feature_engineered.csv")
```

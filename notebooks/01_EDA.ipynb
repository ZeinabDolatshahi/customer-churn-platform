# Business Insights – Employee Attrition Analysis

## Class Distribution

- The dataset contains **1,470 employees**, of which **1,233 (83.88%)** remained with the company and **237 (16.12%)** left the organization.
- The target variable is **imbalanced**, with the majority of employees belonging to the "No Attrition" class.

## Key Observations

- Employee attrition represents approximately **one out of every six employees**.
- The imbalance suggests that predicting employee turnover is a challenging classification problem, as machine learning models may become biased toward the majority class.
- During model development, techniques such as **class weighting** or **resampling methods** (e.g., SMOTE) may be considered to improve the prediction of the minority class.

## Business Implication

Although employee attrition affects a relatively small proportion of the workforce, it represents a significant business concern. Identifying employees at risk of leaving can help organizations implement targeted retention strategies, reduce recruitment costs, and improve workforce stability.

### Business Implications

The numerical variables capture important employee characteristics that may influence attrition, including age, income, work experience, and commuting distance. Constant variables and unique identifiers will be removed during feature engineering, while the remaining numerical features will be further evaluated for their predictive importance in the machine learning model.
### Boxplot Analysis

- The boxplots reveal that several numerical variables contain potential outliers, particularly **MonthlyIncome**, **MonthlyRate**, **DailyRate**, **TotalWorkingYears**, and **YearsAtCompany**.

- Rating variables, including **JobSatisfaction**, **EnvironmentSatisfaction**, **RelationshipSatisfaction**, and **WorkLifeBalance**, exhibit limited variability because they are measured on discrete ordinal scales.

### Business Implications

The detected outliers are likely to reflect genuine employee differences rather than data quality issues. Therefore, they will be retained during exploratory analysis and evaluated further during feature engineering and model development. Constant variables (**EmployeeCount** and **StandardHours**) and the identifier (**EmployeeNumber**) will be removed before training the machine learning model.

```python
# ==========================================================
# 6. CORRELATION ANALYSIS
# ==========================================================

correlation_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(14,10))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()

# Strongest correlations
corr = correlation_matrix.unstack()

corr = corr.sort_values(
    ascending=False,
    key=lambda x: abs(x)
)

display(corr.head(20))
```

# Business Insights – Correlation Analysis

## Correlation Between Numerical Variables

The correlation matrix was used to examine the relationships among the numerical features and to identify potential multicollinearity before model development.

### Key Observations

- Most numerical variables exhibit **weak to moderate correlations**, indicating that they capture different aspects of employee characteristics.
- **MonthlyIncome** and **JobLevel** show a **very strong positive correlation** (approximately **0.95**), suggesting that employees at higher job levels generally earn higher salaries.
- **TotalWorkingYears** is strongly correlated with **JobLevel** (≈ **0.78**) and **MonthlyIncome** (≈ **0.77**), indicating that employees with more work experience tend to hold higher-level and better-paid positions.
- **Age** has a moderate positive correlation with **TotalWorkingYears** (≈ **0.68**), reflecting that older employees generally have more work experience.
- Variables related to tenure, including **YearsAtCompany**, **YearsInCurrentRole**, **YearsSinceLastPromotion**, and **YearsWithCurrManager**, are positively correlated because they measure related aspects of an employee's career progression.
- Most satisfaction-related variables (e.g., **EnvironmentSatisfaction**, **JobSatisfaction**, **RelationshipSatisfaction**, and **WorkLifeBalance**) show weak correlations with other numerical features, suggesting they provide independent information.

### Business Implications

The correlation analysis indicates that the dataset has **limited multicollinearity**, making it suitable for predictive modeling. While some variables describing employee experience and compensation are highly correlated, they reflect meaningful business relationships rather than data quality issues. In the next stage, machine learning models and SHAP analysis will be used to determine which features contribute most to employee attrition.
